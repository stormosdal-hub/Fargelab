# Notatbok: kom i gang

En notatbok består av celler. Kjør en celle med **Shift+Enter**. Variablene du lager i én celle, kan du bruke i de neste, akkurat som i Google Colab.

- **Shift+Enter** kjører cellen og går til neste
- **Ctrl+Enter** kjører cellen og blir stående
- **Alt+Enter** kjører cellen og lager en ny under
- Dobbeltklikk på en tekstcelle for å redigere den

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("eksempel.png")   # fra mappen bilder/
img.shape
```

## Gråtone og terskel

`img` fra cellen over finnes fortsatt. Endre `thresh` og kjør cellen på nytt.

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
thresh = 120
_, binary = cv2.threshold(gray, thresh, 255, cv2.THRESH_BINARY)
cv2.imshow(f"thresh = {thresh}", binary)
```

## Histogram

Figurer fra matplotlib vises rett under cellen.

```python
plt.figure(figsize=(7, 3))
plt.hist(gray.ravel(), bins=256, range=(0, 256), color="gray")
plt.axvline(thresh, color="red", label=f"thresh = {thresh}")
plt.legend()
plt.title("Gråtoner i eksempel.png")
plt.show()
```
